```python
# write a basic crud file for python
# Here's a basic CRUD (Create, Read, Update, Delete) implementation in Python using a simple in-memory list to store data. This example demonstrates how to manage a collection of items (in this case, a list of dictionaries representing users) with basic CRUD operations.

class UserManager:
    def __init__(self):
        self.users = []

    def create_user(self, user_id, name):
        user = {'id': user_id, 'name': name}
        self.users.append(user)
        return user

    def read_user(self, user_id):
        for user in self.users:
            if user['id'] == user_id:
                return user
        return None

    def update_user(self, user_id, new_name):
        for user in self.users:
            if user['id'] == user_id:
                user['name'] = new_name
                return user
        return None

    def delete_user(self, user_id):
        for i, user in enumerate(self.users):
            if user['id'] == user_id:
                del self.users[i]
                return True
        return False

# Example usage
user_manager = UserManager()
user_manager.create_user(1, "John Doe")
user_manager.create_user(2, "Jane Smith")

print(user_manager.read_user(1))  # Read a user
user_manager.update_user(1, "New Name")  # Update a user
user_manager.delete_user(2)  # Delete a user
```
